```python
# ============================================
# 🚗 CAR DEKHO DATASET - PREPROCESSING
# ============================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# --------------------------------------------
# 1. Load Dataset
# --------------------------------------------
file_path = "/kaggle/input/datasets/sartajhuccain/car-detail/CAR DETAILS FROM CAR DEKHO (4).csv"

df = pd.read_csv(file_path)

print("Original Shape:", df.shape)

# --------------------------------------------
# 2. Remove Duplicate Rows
# --------------------------------------------
df = df.drop_duplicates()

# --------------------------------------------
# 3. Handle Missing Values
# --------------------------------------------
print("\nMissing Values Before:")
print(df.isnull().sum())

# Numerical columns -> median
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns -> mode
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values After:")
print(df.isnull().sum())

# --------------------------------------------
# 4. Separate Features and Target
# --------------------------------------------
target = "selling_price"

X = df.drop(columns=[target])
y = df[target]

# --------------------------------------------
# 5. Select Columns
# --------------------------------------------

# Categorical columns
categorical_features = [
    "fuel",
    "seller_type",
    "transmission",
    "owner"
]

# Numerical columns
numerical_features = [
    "year",
    "km_driven"
]

# --------------------------------------------
# 6. Handle Car Name
# --------------------------------------------
# 'name' contains many different car models.
# One-hot encoding thousands of names can make
# the model unnecessarily large, so we remove it.

if "name" in X.columns:
    X = X.drop(columns=["name"])

# --------------------------------------------
# 7. Train-Test Split
# --------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)

# --------------------------------------------
# 8. Preprocessing Pipeline
# --------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ]
)

# --------------------------------------------
# 9. Transform Data
# --------------------------------------------
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("\nProcessed Training Shape:", X_train_processed.shape)
print("Processed Testing Shape:", X_test_processed.shape)

print("\n✅ Preprocessing Completed Successfully!")
```

```python
# ============================================
# 🚗 XGBOOST REGRESSION - CAR PRICE PREDICTION
# ============================================

from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# --------------------------------------------
# 1. Create XGBoost Model
# --------------------------------------------

model = XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    min_child_weight=2,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

# --------------------------------------------
# 2. Train Model
# --------------------------------------------

print("🚀 Training XGBoost Regression Model...")

model.fit(
    X_train_processed,
    y_train
)

print("✅ Model Training Completed!")

# --------------------------------------------
# 3. Prediction
# --------------------------------------------

y_pred = model.predict(X_test_processed)

# --------------------------------------------
# 4. Evaluation
# --------------------------------------------

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("\n" + "=" * 50)
print("📊 XGBoost Regression Results")
print("=" * 50)

print(f"MAE  : {mae:,.2f}")
print(f"MSE  : {mse:,.2f}")
print(f"RMSE : {rmse:,.2f}")
print(f"R²   : {r2:.4f}")

print("=" * 50)
```

```python
# ==========================================================
# 🚗 XGBOOST REGRESSION
# VISUALIZATION + EVALUATION
# ==========================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ----------------------------------------------------------
# 1. Predictions
# ----------------------------------------------------------
y_pred = model.predict(X_test_processed)

# ----------------------------------------------------------
# 2. Evaluation Metrics
# ----------------------------------------------------------
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("=" * 60)
print("📊 XGBoost Regression Evaluation")
print("=" * 60)

print(f"MAE  : {mae:,.2f}")
print(f"MSE  : {mse:,.2f}")
print(f"RMSE : {rmse:,.2f}")
print(f"R² Score : {r2:.4f}")

# ----------------------------------------------------------
# 3. Evaluation Metrics Bar Chart
# ----------------------------------------------------------
metrics = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R² Score"],
    "Value": [mae, mse, rmse, r2]
})

plt.figure(figsize=(10, 6))
sns.barplot(data=metrics, x="Metric", y="Value")
plt.title("XGBoost Regression Evaluation Metrics")
plt.xlabel("Evaluation Metric")
plt.ylabel("Value")
plt.show()

# ----------------------------------------------------------
# 4. Actual vs Predicted Price
# ----------------------------------------------------------
plt.figure(figsize=(10, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.6
)

# Perfect prediction line
min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.title("Actual vs Predicted Selling Price")
plt.xlabel("Actual Selling Price")
plt.ylabel("Predicted Selling Price")
plt.show()

# ----------------------------------------------------------
# 5. Residual Plot
# ----------------------------------------------------------
residuals = y_test - y_pred

plt.figure(figsize=(10, 6))

plt.scatter(
    y_pred,
    residuals,
    alpha=0.6
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title("Residual Plot")
plt.xlabel("Predicted Selling Price")
plt.ylabel("Residual (Actual - Predicted)")
plt.show()

# ----------------------------------------------------------
# 6. Residual Distribution
# ----------------------------------------------------------
plt.figure(figsize=(10, 6))

sns.histplot(
    residuals,
    kde=True,
    bins=30
)

plt.title("Residual Distribution")
plt.xlabel("Residual")
plt.ylabel("Frequency")
plt.show()

# ----------------------------------------------------------
# 7. Actual vs Predicted Line Plot
# ----------------------------------------------------------
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
}).reset_index(drop=True)

# First 100 samples for better visualization
comparison_100 = comparison.head(100)

plt.figure(figsize=(14, 6))

plt.plot(
    comparison_100["Actual"],
    label="Actual"
)

plt.plot(
    comparison_100["Predicted"],
    label="Predicted"
)

plt.title("Actual vs Predicted Selling Price - First 100 Samples")
plt.xlabel("Sample")
plt.ylabel("Selling Price")
plt.legend()
plt.show()

# ----------------------------------------------------------
# 8. Prediction Error
# ----------------------------------------------------------
comparison["Error"] = (
    comparison["Actual"] - comparison["Predicted"]
)

comparison["Absolute Error"] = (
    abs(comparison["Error"])
)

print("\n🔹 First 10 Predictions:")
display(comparison.head(10))

# ----------------------------------------------------------
# 9. Model Performance Summary
# ----------------------------------------------------------
print("\n" + "=" * 60)
print("✅ FINAL MODEL PERFORMANCE")
print("=" * 60)

print(f"Mean Absolute Error  : {mae:,.2f}")
print(f"Mean Squared Error   : {mse:,.2f}")
print(f"Root Mean Squared Error : {rmse:,.2f}")
print(f"R² Score             : {r2:.4f}")

print("=" * 60)
```
